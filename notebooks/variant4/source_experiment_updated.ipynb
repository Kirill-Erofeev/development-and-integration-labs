# Обновлённый эксперимент: безопасный паспорт модели

Endpoint `/model-info` сообщает только сведения, полезные клиенту и безопасные для публикации.


```python
from pathlib import Path
import sys

project_root = next(
    parent for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / "src" / "model_service.py").is_file()
)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.model_service import ModelService

service = ModelService.load(project_root / "models" / "model.pkl")
```

```python
passport = service.model_info()
assert passport == {
    "model_type": "LogisticRegression",
    "classes": ["setosa", "versicolor", "virginica"],
    "feature_count": 4,
    "model_loaded": True,
}
passport
```
