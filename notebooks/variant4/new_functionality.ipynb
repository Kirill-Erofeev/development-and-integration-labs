# Новый эксперимент: request ID и событие прогноза

UUID связывает действие на странице с ответом API и записью в журнале.


```python
from pathlib import Path
import sys

project_root = next(
    parent for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / "src" / "model_service.py").is_file()
)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from uuid import UUID
from src.request_id import resolve_request_id, prediction_event
```

```python
new_id = resolve_request_id()
assert str(UUID(new_id)) == new_id
assert UUID(new_id).version == 4
assert resolve_request_id(new_id.upper()) == new_id
assert UUID(resolve_request_id("invalid")).version == 4
prediction_event(new_id, "setosa")
```

## Перенос в проект

Прочитайте `X-Request-ID` в HTTP-слое, проверьте UUID и передайте идентификатор в прикладную функцию и лог. Не помещайте в лог все необработанные данные запроса.

Выбран единый контракт: отсутствующий или неверный UUID заменяется новым UUIDv4. Функции импортируются из рабочего модуля; API отправляет событие в стандартный поток ошибок через logging.
