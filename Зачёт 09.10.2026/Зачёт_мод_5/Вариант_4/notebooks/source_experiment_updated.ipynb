# Обновлённый эксперимент: безопасный паспорт модели

Endpoint `/model-info` сообщает только сведения, полезные клиенту и безопасные для публикации.


```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

iris = load_iris()
model = LogisticRegression(max_iter=300, random_state=42).fit(iris.data, iris.target)
```

```python
def model_info(model, class_names):
    """Не раскрывает путь к файлу, параметры окружения и сериализованный объект."""
    return {
        'model_type': type(model).__name__,
        'classes': list(class_names),
        'feature_count': int(model.n_features_in_),
        'model_loaded': True,
    }

passport = model_info(model, iris.target_names)
assert passport['feature_count'] == 4
passport
```
