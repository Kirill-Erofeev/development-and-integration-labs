# Новый эксперимент: request ID и событие прогноза

UUID связывает действие на странице с ответом API и записью в журнале.


```python
from uuid import UUID, uuid4

def resolve_request_id(value=None):
    if value is None:
        return str(uuid4())
    return str(UUID(value))  # ValueError для некорректного заголовка
```

```python
def prediction_event(request_id, class_name):
    """Минимальная JSON-совместимая запись для структурированного лога."""
    return {'event': 'prediction_completed', 'request_id': request_id, 'class_name': class_name}

new_id = resolve_request_id()
assert str(UUID(new_id)) == new_id
prediction_event(new_id, 'setosa')
```

## Перенос в проект

Прочитайте `X-Request-ID` в HTTP-слое, проверьте UUID и передайте идентификатор в прикладную функцию и лог. Не помещайте в лог все необработанные данные запроса.
